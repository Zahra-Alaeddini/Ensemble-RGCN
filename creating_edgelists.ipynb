```python
import pandas as pd

disease_metabolic = pd.read_csv('disease_metabolite.csv')
edgelist = []

for index, row in disease_metabolic.iterrows():
    disease = row['disease_info']
    metabolite = row['pubchem_id']
    alteration = row['alteration']
    edgelist.append([disease, metabolite, alteration])

edgelist_df = pd.DataFrame(edgelist, columns=['Source', 'Target', 'Relationship'])
edgelist_df.to_csv('disease_metabolite_edgelist.csv', index=False)
```

```python
# Do the same for the source column

import pandas as pd

def replace_target_with_metabolic(dataset1_path, dataset2_path, output_path):
    df1 = pd.read_csv(dataset1_path)  
    df2 = pd.read_csv(dataset2_path)  

    id_to_chemical = pd.Series(df1['metabolic'].values, index=df1['pubchem_id']).to_dict()

    def replace_target(row):
        try:
            target_value = str(row['Target'])
            if target_value in id_to_chemical:
                return id_to_chemical[target_value]
        except ValueError:
            pass
        return row['Target']

    df2['Target'] = df2.apply(replace_target, axis=1)
    df2.to_csv(output_path, index=False)
    print(f"Modified dataset saved to {output_path}")


dataset1_path = 'disease_metabolite.csv'   # Use main data before creating edge lists
dataset2_path = 'All-predicted-Links.csv'  
output_path = 'New-predicted-links.csv'  # The path where the modified dataset will be saved

replace_target_with_metabolic(dataset1_path, dataset2_path, output_path)
```

```python
# Do the same for the source column

def replace_target_with_gene(dataset1_path, dataset2_path, output_path):
    df1 = pd.read_csv(dataset1_path)  
    df2 = pd.read_csv(dataset2_path) 
    id_to_gene = pd.Series(df1['gene'].values, index=df1['gene_id']).to_dict()

    def replace_target(row):
        try:
            target_value = str(row['Target'])
            if target_value in id_to_gene:
                return id_to_gene[target_value]
        except ValueError:
            pass
        return row['Target']

    df2['Target'] = df2.apply(replace_target, axis=1)

    df2.to_csv(output_path, index=False)
    print(f"Modified dataset saved to {output_path}")

dataset1_path = 'metabolite_gene.csv'   # Use main data that used for creating edge lists
dataset2_path = 'New-predicted-Links.csv'  
output_path = 'Final-Predicted-Links.csv'  

replace_target_with_gene(dataset1_path, dataset2_path, output_path)
```

```python
# this code use for finding various element has relation with a specific entity

import pandas as pd
def search_in_csv(file_path, search_string, output_file):
    df = pd.read_csv(file_path)

    matches = df[df.apply(lambda row: row.astype(str).str.contains(search_string, case=False).any(), axis=1)]

    if not matches.empty:
        print("Rows containing the search string:")
        print(matches)
        matches.to_csv(output_file, index=False)
        print(f"Matching rows saved to {output_file}")
    else:
        print("No matches found.")
        matches.to_csv(output_file, index=False)
        print(f"Empty file saved to {output_file} (No matches found)")

input_path = 'Final-Predicted-Links.csv'  
output_path = 'Filtered_Results.csv'     

search_string = input("Enter a string to search in the modified dataset: ")
search_in_csv(input_path, search_string, output_path)
```
